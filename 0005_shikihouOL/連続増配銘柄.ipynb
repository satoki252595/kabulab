```python
import csv
from bs4 import BeautifulSoup

html_text = """
<tbody> ... （ここに頂いたHTML全文） ... </tbody>
"""

# BeautifulSoupで解析
soup = BeautifulSoup(html_text, "html.parser")

# tbodyタグを取得
tbody = soup.find("tbody")
if not tbody:
    raise ValueError("tbodyが見つかりません。")

# trタグ（各行）をすべて取得
rows = tbody.find_all("tr")

# CSV書き込み用のファイルを開く（書き込みモード）
# newline='', encoding='utf-8' は環境に合わせて設定してください
with open("output.csv", mode="w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)

    # ---- 1) ヘッダー行の定義 ----
    # 今回のHTMLを見ると、td列が18個あるようですので、それぞれ対応する見出しを定義します。
    # 必要に応じてカスタマイズしてご利用ください。
    header = ["MIC001", "MIC004", "MIC005", "MIC008", "MI0001", "MI0006", "MCE0040", "MI0211", "MI0212", "MI0959", "MI1070", "MI0963"]
    writer.writerow(header)

    # ---- 2) 各行（tr）ごとにデータを取得してCSVへ書き込み ----
    for tr in rows:
        # tdタグ（セル）をすべて取得
        tds = tr.find_all("td")
        # tdsが空なら次のtrへ（空行やコメント行などをスキップ）
        if not tds:
            continue
        
        # 各セルのテキストをstrip()で取得してリストにまとめる
        row_data = [td.get_text(strip=True) for td in tds]

        # CSVへ1行分を書き込み
        writer.writerow(row_data)

print("書き込みが完了しました。")
```
