```python
import csv
from bs4 import BeautifulSoup

# （2）BeautifulSoup でパース
soup = BeautifulSoup(html_snippet, "html.parser")

# （3）取得したい列定義
#     業種分類カラム (MIC006, MIC007, MIC008) を含めています
columns = [
    ("銘柄コード", "td.vgt-left-align.-MIC001 p"),               # 銘柄コード
    ("銘柄名", "td.vgt-left-align.-fix.-MIC004 a"),          # 銘柄名
    ("市場", "td.vgt-left-align.-MIC005.-checkbox span"),  # 市場

    # ★ 重要: 業種の3つ。クラス指定に -checkbox も含め、ドットでつないでいる
    ("", "td.vgt-left-align.-MIC006.-checkbox span"),  # 業種大分類
    ("業種大分類", "td.vgt-left-align.-MIC007.-checkbox span"),  # 業種中分類
    ("業種小分類", "td.vgt-left-align.-MIC008.-checkbox span"),  # 業種小分類
    ("時価総額", "td.vgt-right-align.-MI0002 span"),  
    ("株価", "td.vgt-right-align.-MI0001 span"),  # 株価など
    ("変動率", "td.vgt-right-align.-MI0006 div"),   # 変動率
    ("配当利回り", "td.vgt-right-align.-MI0979 span"),  # 新規追加1
    ("営業利益率", "td.vgt-right-align.-MI0963 span"),  # 新規追加2
    ("特色", "td.vgt-left-align.-MI1070 span"),   # 企業情報
]

# （4）<tbody> 配下の <tr> を取得
rows = soup.select("tbody tr")

# （5）1行ずつパースしてデータを取得
all_data = []
for tr in rows:
    row_data = {}
    for col_name, selector in columns:
        elem = tr.select_one(selector)
        text_value = elem.get_text(strip=True) if elem else ""
        row_data[col_name] = text_value
    all_data.append(row_data)

# （6）CSVへ出力（例: 'output.csv'）
with open("output1.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    # ヘッダー
    writer.writerow([col[0] for col in columns])
    # 各行データ
    for row_data in all_data:
        writer.writerow([row_data[col[0]] for col in columns])

print("CSV 出力完了: output.csv")
```
