```python
import yfinance as yf
import pandas as pd
import re
import urllib.request
import tabula


def getStockCodeDataFrame() -> pd.core.frame.DataFrame:
    '''
    東証の上場銘柄コード一覧をdf形式で出力する。
    URL = 'https://www.jpx.co.jp/markets/statistics-equities/misc/01.html'
    '''

    URL = 'https://www.jpx.co.jp/markets/statistics-equities/misc/01.html'

    response = urllib.request.urlopen(URL).read().decode("utf-8")
    string_html = re.findall('<a href=\".+?\.xls\"', response)
    url_list = []
    for i in string_html:
        j = i.lstrip('<a href=\"')
        k = j.rstrip('\"')
        url_list.append('https://www.jpx.co.jp'+k)
    url = url_list[0]

    # 国内株式のみ抽出

    df = pd.read_excel(url)
    df = df[(df.iloc[:, 3] == 'プライム（内国株式）') | (df.iloc[:, 3] ==
                                               'スタンダード（内国株式）') | (df.iloc[:, 3] == 'グロース（内国株式）')]

    # 列名の変更
    df.columns = ['date', 'code', 'office_name', 'market_class', 'industry_detail_code',
                  'industry_detail', 'industry_code', 'industry', 'scale_code', 'scale_class']

    # 'date'列をdatetime->strへ型変換
    df['date'] = df['date'].astype(str)
    df['code'] = df['code'].astype(str)
    df['industry_detail_code'] = df['industry_detail_code'].astype(str)
    df['industry_code'] = df['industry_code'].astype(str)
    df['scale_code'] = df['scale_code'].astype(str)

    return df


def get_stock_data(ticker):
    stock = yf.Ticker(ticker)
    hist = stock.history(period="3mo")
    return hist


def is_perfect_order(hist) -> bool:
    ma5 = hist['Close'].rolling(window=5).mean()
    ma10 = hist['Close'].rolling(window=10).mean()
    ma20 = hist['Close'].rolling(window=20).mean()
    ma60 = hist['Close'].rolling(window=60).mean()
    return (ma5.iloc[-1] > ma10.iloc[-1] > ma20.iloc[-1] > ma60.iloc[-1])


def screen_stocks(tickers):
    screened_stocks = []
    for ticker in tickers:
        hist = get_stock_data(ticker)
        if len(hist) < 60:
            continue
        avg_volume_1w = hist['Volume'].rolling(window=5).mean().iloc[-1]
        avg_volume_1m = hist['Volume'].rolling(window=20).mean().iloc[-1]
        stock = yf.Ticker(ticker)
        market_cap = stock.info.get('marketCap')

        if avg_volume_1w > 1.5 * avg_volume_1m and is_perfect_order(hist) and market_cap < 50000000000:
            screened_stocks.append(ticker)
    return screened_stocks


df = getStockCodeDataFrame()
code_list = [code + '.T' for code in df['code'].tolist()]

screened_stocks = screen_stocks(code_list)
print("Screened Stocks:", screened_stocks)
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
from datetime import datetime, timedelta


def get_stock_financial_info(ticker_symbol):
    """
    指定した銘柄の財務情報を取得する関数

    Parameters:
    ticker_symbol (str): 銘柄コード（例：'7203.T'）

    Returns:
    dict: 財務情報を含む辞書
    """
    # Tickerオブジェクトの作成
    ticker = yf.Ticker(ticker_symbol)

    try:
        # 基本情報の取得
        info = ticker.info

        # 財務諸表の取得
        financials = ticker.financials

        # 結果を格納する辞書
        result = {
            '基本情報': {
                '株式発行数': info.get('sharesOutstanding', 'データなし'),
                '時価総額': info.get('marketCap', 'データなし'),
                'ROE': info.get('returnOnEquity', 'データなし'),
                '事業内容': info.get('longBusinessSummary', 'データなし')

            },
            '財務情報': None,
            'データ取得状況': {
                '基本情報': True,
                '財務情報': False
            }
        }

        # 財務情報の取得と計算
        financial_data = []

        if not financials.empty:
            # 各期間のデータを処理
            for date in financials.columns:
                period_data = {}
                period_data['期間'] = date.strftime('%Y-%m-%d')

                # 売上高の取得
                try:
                    revenue = financials.loc['Total Revenue', date]
                    period_data['売上高'] = revenue if not pd.isna(
                        revenue) else None
                except KeyError:
                    period_data['売上高'] = None

                # 営業利益の取得
                try:
                    operating_income = financials.loc['Operating Income', date]
                    period_data['営業利益'] = operating_income if not pd.isna(
                        operating_income) else None
                except KeyError:
                    period_data['営業利益'] = None

                # 営業利益率の計算
                if period_data['売上高'] and period_data['営業利益']:

                    period_data['営業利益率'] = round(
                        (period_data['営業利益'] / period_data['売上高'] * 100), 2)
                else:
                    period_data['営業利益率'] = None

                # 有効なデータが1つでもあれば追加
                if any(v is not None for v in period_data.values()):
                    financial_data.append(period_data)

            if financial_data:
                result['財務情報'] = pd.DataFrame(financial_data)
                result['データ取得状況']['財務情報'] = True

        return result

    except Exception as e:
        print(f"データ取得中にエラーが発生しました: {str(e)}")
        return None


def format_number(value):
    """
    数値を見やすい形式にフォーマットする関数
    """
    if pd.isna(value) or value == 'データなし':
        return 'データなし'
    elif isinstance(value, (int, float)):
        if value >= 1000000000:  # 10億以上
            return f"¥{value/1000000000:.2f}B"
        elif value >= 1000000:  # 100万以上
            return f"¥{value/1000000:.2f}M"
        else:
            return f"¥{value:,.0f}"
    return value


def display_financial_info(ticker_symbol):
    """
    財務情報を見やすく表示する関数

    Parameters:
    ticker_symbol (str): 銘柄コード
    """
    result = get_stock_financial_info(ticker_symbol)

    if result is None:
        print("データを取得できませんでした。")
        return

    print(f"\n=== {ticker_symbol} の財務情報 ===\n")

    # 基本情報の表示
    print("【基本情報】")
    if result['データ取得状況']['基本情報']:
        print(f"株式発行数: {format_number(result['基本情報']['株式発行数'])}")
        print(f"時価総額: {format_number(result['基本情報']['時価総額'])}")
        roe = result['基本情報']['ROE']
        if roe is not None:
            roe_percentage = f"{roe * 100:.1f}%"
        else:
            roe_percentage = 'データなし'
        print(f"ROE: {roe_percentage}")
        print(f"事業内容: {result['基本情報']['事業内容']}")
    else:
        print("基本情報を取得できませんでした。")

    print("\n【財務情報】")
    if result['データ取得状況']['財務情報'] and result['財務情報'] is not None:
        # 財務情報のフォーマット
        formatted_df = result['財務情報'].copy()
        for col in ['売上高', '営業利益']:
            if col in formatted_df.columns:
                formatted_df[col] = formatted_df[col].apply(format_number)
        if '営業利益率' in formatted_df.columns:
            formatted_df['営業利益率'] = formatted_df['営業利益率'].apply(
                lambda x: f"{x:.2f}%" if pd.notnull(x) else 'データなし'
            )

        print(formatted_df.to_string(index=False))
    else:
        print("財務情報を取得できませんでした。")


# 使用例
if __name__ == "__main__":
    # 日本株の例（トヨタ自動車）
    ticker_symbol = '9927.T'
    display_financial_info(ticker_symbol)
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
from datetime import datetime
from typing import Dict, Any, Optional, Tuple

class StockFinancialAnalyzer:
    """株式の財務情報を包括的に分析するクラス"""
    
    # 財務諸表項目の和英対応辞書
    FINANCIAL_TERMS = {
        # 損益計算書項目
        'Total Revenue': '売上高',
        'Operating Income': '営業利益',
        'Net Income': '当期純利益',
        'Gross Profit': '売上総利益',
        'Operating Expense': '営業費用',
        'Interest Expense': '支払利息',
        'EBITDA': 'EBITDA(償却前営業利益)',
        'Cost of Revenue': '売上原価',
        'Research Development': '研究開発費',
        'Net Income Common Stockholders': '株主帰属当期純利益',
        
        # 貸借対照表項目
        'Total Assets': '総資産',
        'Total Liabilities': '負債合計',
        'Total Stockholder Equity': '株主資本',
        'Cash And Cash Equivalents': '現金及び現金同等物',
        'Total Current Assets': '流動資産合計',
        'Total Current Liabilities': '流動負債合計',
        'Long Term Debt': '長期借入金',
        'Accounts Payable': '買掛金',
        'Accounts Receivable': '売掛金',
        'Inventory': '棚卸資産',
        'Property Plant Equipment': '有形固定資産',
        'Retained Earnings': '利益剰余金',
        
        # キャッシュフロー計算書項目
        'Operating Cash Flow': '営業キャッシュフロー',
        'Investing Cash Flow': '投資キャッシュフロー',
        'Financing Cash Flow': '財務キャッシュフロー',
        'Capital Expenditure': '設備投資',
        'Free Cash Flow': 'フリーキャッシュフロー',
        'Net Borrowings': '純借入額',
        'Change In Cash': '現金増減額',
        'Repurchase Of Stock': '自己株式取得',
        'Dividends Paid': '配当金支払額'
    }
    
    def __init__(self, ticker_symbol: str):
        """
        Parameters:
        ticker_symbol (str): 銘柄コード（例：'7203.T'）
        """
        self.ticker_symbol = ticker_symbol
        self.ticker = yf.Ticker(ticker_symbol)
        
        # データ保持用の属性
        self.basic_info: Dict[str, Any] = {}
        self.income_statement: Optional[pd.DataFrame] = None
        self.balance_sheet: Optional[pd.DataFrame] = None
        self.cash_flow: Optional[pd.DataFrame] = None
        self.quarterly_income_statement: Optional[pd.DataFrame] = None
        self.quarterly_balance_sheet: Optional[pd.DataFrame] = None
        self.quarterly_cash_flow: Optional[pd.DataFrame] = None
    
    def _is_valid_dataframe(self, df: Optional[pd.DataFrame]) -> bool:
        """
        データフレームが有効なデータを含んでいるかチェック
        """
        if df is None or df.empty:
            return False
        
        # 全ての値がNaN or 0かチェック
        return not df.fillna(0).astype(bool).sum().sum() == 0
    
    def fetch_all_financial_data(self) -> None:
        """全ての財務データを取得"""
        self._fetch_basic_info()
        self._fetch_financial_statements()
    
    def _fetch_basic_info(self) -> None:
        """基本情報の取得"""
        try:
            info = self.ticker.info
            base_info = {
                'Company Name (企業名)': info.get('longName', 'データなし'),
                'Ticker Symbol (証券コード)': self.ticker_symbol,
                'Industry (業種)': info.get('industry', 'データなし'),
                'Sector (セクター)': info.get('sector', 'データなし'),
                'Shares Outstanding (発行済株式数)': info.get('sharesOutstanding', 'データなし'),
                'Market Cap (時価総額)': info.get('marketCap', 'データなし'),
                'Stock Price (株価)': info.get('currentPrice', 'データなし'),
                'PER (株価収益率)': info.get('trailingPE', 'データなし'),
                'PBR (株価純資産倍率)': info.get('priceToBook', 'データなし'),
                'Dividend Yield (配当利回り)': info.get('dividendYield', 'データなし'),
                '52 Week High (52週高値)': info.get('fiftyTwoWeekHigh', 'データなし'),
                '52 Week Low (52週安値)': info.get('fiftyTwoWeekLow', 'データなし')
            }
            
            # 有効なデータがある項目のみを保持
            self.basic_info = {k: v for k, v in base_info.items() 
                             if v not in ['データなし', None, 0, float('nan')]}
            
        except Exception as e:
            print(f"基本情報の取得中にエラーが発生しました: {str(e)}")
    
    def _fetch_financial_statements(self) -> None:
        """財務諸表の取得"""
        try:
            # 年次財務諸表の取得と検証
            temp_income = self.ticker.financials
            temp_balance = self.ticker.balance_sheet
            temp_cash_flow = self.ticker.cashflow
            
            # 有効なデータがある場合のみ保持
            self.income_statement = temp_income if self._is_valid_dataframe(temp_income) else None
            self.balance_sheet = temp_balance if self._is_valid_dataframe(temp_balance) else None
            self.cash_flow = temp_cash_flow if self._is_valid_dataframe(temp_cash_flow) else None
            
            # 四半期財務諸表の取得と検証
            temp_q_income = self.ticker.quarterly_financials
            temp_q_balance = self.ticker.quarterly_balance_sheet
            temp_q_cash_flow = self.ticker.quarterly_cashflow
            
            # 有効なデータがある場合のみ保持
            self.quarterly_income_statement = temp_q_income if self._is_valid_dataframe(temp_q_income) else None
            self.quarterly_balance_sheet = temp_q_balance if self._is_valid_dataframe(temp_q_balance) else None
            self.quarterly_cash_flow = temp_q_cash_flow if self._is_valid_dataframe(temp_q_cash_flow) else None
            
        except Exception as e:
            print(f"財務諸表の取得中にエラーが発生しました: {str(e)}")
    
    def _format_financial_value(self, value: Any) -> str:
        """財務数値のフォーマット"""
        if pd.isna(value) or value == 'データなし':
            return 'データなし'
        elif isinstance(value, (int, float)):
            if abs(value) >= 1000000000000:  # 1兆以上
                return f"¥{value/1000000000000:.2f}T"
            elif abs(value) >= 1000000000:  # 10億以上
                return f"¥{value/1000000000:.2f}B"
            elif abs(value) >= 1000000:  # 100万以上
                return f"¥{value/1000000:.2f}M"
            elif isinstance(value, float):
                return f"{value:.2f}"
            else:
                return f"{value:,}"
        return str(value)
    
    def _format_percentage(self, value: Any) -> str:
        """パーセンテージのフォーマット"""
        if pd.isna(value) or value == 'データなし':
            return 'データなし'
        elif isinstance(value, (int, float)):
            return f"{value:.2f}%"
        return str(value)
    
    def display_basic_info(self) -> None:
        """基本情報の表示"""
        if not self.basic_info:
            print("\n=== Basic Information (基本情報) ===")
            print("基本情報を取得できませんでした。")
            return
            
        print("\n=== Basic Information (基本情報) ===")
        for key, value in self.basic_info.items():
            formatted_value = self._format_financial_value(value)
            print(f"{key}: {formatted_value}")
    
    def format_financial_statement(self, df: Optional[pd.DataFrame]) -> Optional[pd.DataFrame]:
        """財務諸表のフォーマット"""
        if not self._is_valid_dataframe(df):
            return None
            
        # データフレームのコピーを作成
        formatted_df = df.copy()
        
        # インデックスを和英併記に変更
        new_index = [f"{idx} ({self.FINANCIAL_TERMS.get(idx, '不明')})" 
                    if idx in self.FINANCIAL_TERMS else idx 
                    for idx in formatted_df.index]
        formatted_df.index = new_index
        
        # 数値をフォーマット
        formatted_df = formatted_df.applymap(self._format_financial_value)
        
        return formatted_df
    
    def display_financial_statements(self, period: str = 'annual') -> None:
        """財務諸表の表示
        
        Parameters:
        period (str): 'annual' or 'quarterly'
        """
        if period == 'annual':
            statements = [
                ('Income Statement (損益計算書)', self.income_statement),
                ('Balance Sheet (貸借対照表)', self.balance_sheet),
                ('Cash Flow Statement (キャッシュフロー計算書)', self.cash_flow)
            ]
        else:
            statements = [
                ('Quarterly Income Statement (四半期損益計算書)', self.quarterly_income_statement),
                ('Quarterly Balance Sheet (四半期貸借対照表)', self.quarterly_balance_sheet),
                ('Quarterly Cash Flow Statement (四半期キャッシュフロー計算書)', self.quarterly_cash_flow)
            ]
        
        for title, df in statements:
            print(f"\n=== {title} ===")
            if self._is_valid_dataframe(df):
                formatted_df = self.format_financial_statement(df)
                if formatted_df is not None:
                    print(formatted_df.to_string())
            else:
                print(f"{title}のデータを取得できませんでした。")
    
    def calculate_financial_metrics(self) -> Optional[pd.DataFrame]:
        """財務指標の計算"""
        if not self._is_valid_dataframe(self.income_statement) or \
           not self._is_valid_dataframe(self.balance_sheet):
            return None
            
        try:
            metrics = pd.DataFrame()
            
            # 売上高成長率 (Revenue Growth Rate)
            if 'Total Revenue' in self.income_statement.index:
                revenue = self.income_statement.loc['Total Revenue']
                metrics.loc['Revenue Growth Rate (売上高成長率)', :] = revenue.pct_change(-1) * 100
            
            # 営業利益率 (Operating Margin)
            if 'Operating Income' in self.income_statement.index and \
               'Total Revenue' in self.income_statement.index:
                operating_income = self.income_statement.loc['Operating Income']
                revenue = self.income_statement.loc['Total Revenue']
                metrics.loc['Operating Margin (営業利益率)', :] = (operating_income / revenue * 100)
            
            # ROE (Return on Equity)
            if 'Net Income' in self.income_statement.index and \
               'Total Stockholder Equity' in self.balance_sheet.index:
                net_income = self.income_statement.loc['Net Income']
                equity = self.balance_sheet.loc['Total Stockholder Equity']
                metrics.loc['ROE (株主資本利益率)', :] = (net_income / equity * 100)
            
            # 流動比率 (Current Ratio)
            if 'Total Current Assets' in self.balance_sheet.index and \
               'Total Current Liabilities' in self.balance_sheet.index:
                current_assets = self.balance_sheet.loc['Total Current Assets']
                current_liabilities = self.balance_sheet.loc['Total Current Liabilities']
                metrics.loc['Current Ratio (流動比率)', :] = (current_assets / current_liabilities * 100)
            
            # 有効なデータがある場合のみ返す
            return metrics if self._is_valid_dataframe(metrics) else None
            
        except Exception as e:
            print(f"財務指標の計算中にエラーが発生しました: {str(e)}")
            return None
    
    def display_financial_metrics(self) -> None:
        """財務指標の表示"""
        metrics = self.calculate_financial_metrics()
        if self._is_valid_dataframe(metrics):
            print("\n=== Financial Metrics (財務指標) ===")
            formatted_metrics = metrics.applymap(self._format_percentage)
            print(formatted_metrics.to_string())
        else:
            print("\n=== Financial Metrics (財務指標) ===")
            print("財務指標を計算できませんでした。")

def analyze_stock(ticker_symbol: str) -> None:
    """株式の財務分析を実行して表示する関数"""
    print(f"\n株式コード {ticker_symbol} の分析を開始します...")
    
    analyzer = StockFinancialAnalyzer(ticker_symbol)
    analyzer.fetch_all_financial_data()
    
    # 基本情報の表示
    analyzer.display_basic_info()
    
    # 年次財務諸表の表示
    print("\n=== Annual Financial Statements (年次財務諸表) ===")
    analyzer.display_financial_statements(period='annual')
    
    # 四半期財務諸表の表示
    print("\n=== Quarterly Financial Statements (四半期財務諸表) ===")
    analyzer.display_financial_statements(period='quarterly')
    
    # 財務指標の表示
    analyzer.display_financial_metrics()
    
    print("\n分析が完了しました。")

# 使用例
if __name__ == "__main__":
    # 日本株の例（トヨタ自動車）
    ticker_symbol = '7203.T'
    analyze_stock(ticker_symbol)
```
